2 Проверка и очистка данных

2.2 Удаление неинформативных строк

```python
stainless_df = stainless_df[
    stainless_df['UTS (MPa)'].notna()
]
```

```python
carbon_df = carbon_df[
    carbon_df['UTS (MPa)'].notna()
]
```

2.3 Приведение числовых типов

```python
numeric_cols_carbon = carbon_df.columns.drop(['SAE Grade', 'Conditions'])
numeric_cols_stainless = stainless_df.columns.drop(['SAE Grade', 'Conditions'])

carbon_df[numeric_cols_carbon] = carbon_df[numeric_cols_carbon].apply(
    pd.to_numeric, errors='coerce'
)

stainless_df[numeric_cols_stainless] = stainless_df[numeric_cols_stainless].apply(
    pd.to_numeric, errors='coerce'
)

```

Преобразование Min / Max → среднее значение

```python
elements = ['C', 'Mn', 'Si', 'Ni', 'Cr', 'Mo', 'Ti']

for el in elements:
    carbon_df[el] = carbon_df[[f'{el} (Min)', f'{el} (Max)']].mean(axis=1)

carbon_df[
    ['SAE Grade', 'Conditions'] +
    [f'{el} (Min)' for el in elements] +
    [f'{el} (Max)' for el in elements] +
    elements
].head()
```

```python
elements = ['C', 'Mn', 'Si', 'Ni', 'Cr', 'Mo', 'Ti']

for el in elements:
    stainless_df[el] = stainless_df[[f'{el} (Min)', f'{el} (Max)']].mean(axis=1)

stainless_df['N'] = pd.to_numeric(stainless_df['N'], errors='coerce')

stainless_df[
    ['SAE Grade', 'Conditions'] +
    [f'{el} (Min)' for el in elements] +
    [f'{el} (Max)' for el in elements] +
    elements +
    ['N']
].head()
```

4 Анализ корреляций

4.1 Корреляции для углеродистых сталей

```python
carbon_features = [
    'C', 'Mn',
    'UTS (MPa)', 'YS (MPa)',
    'Elongation (%)', 'Hardness (HB)'
]
```

Корреляционная матрица

```python
carbon_corr = carbon_df[carbon_features].corr()
carbon_corr
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure()
plt.imshow(carbon_corr)
plt.colorbar()
plt.xticks(range(len(carbon_features)), carbon_features, rotation=45)
plt.yticks(range(len(carbon_features)), carbon_features)
plt.title('Корреляционная матрица для углеродистых сталей')
plt.tight_layout()
plt.show()
```
