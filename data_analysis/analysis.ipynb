# 2 Проверка и очистка данных

## 2.1 Анализ пропущенных значений

```python
carbon_df.isna().sum().sort_values(ascending=False)
```

```python
stainless_df.isna().sum().sort_values(ascending=False)
```

Результаты показывают, что в обоих датасетах присутствует значительное число пропусков, особенно в столбцах, связанных с химическим составом (минимальные и максимальные концентрации элементов).  
Для углеродистых сталей полностью отсутствуют данные по некоторым элементам (например, Ti), а для нержавеющих сталей наблюдается высокая доля пропусков в C (Min), Mn (Min), N и других признаках.  
Это подтверждает необходимость дальнейшей очистки данных и сокращения размерности признакового пространства.

## 2.2 Удаление неинформативных строк

```python
stainless_df = stainless_df[
    stainless_df['UTS (MPa)'].notna()
]

carbon_df = carbon_df[
    carbon_df['UTS (MPa)'].notna()
]
```

Из датасетов удаляются строки, в которых отсутствует значение предела прочности при растяжении (UTS (MPa)).  
Данный показатель является одной из ключевых целевых переменных, поэтому записи без него не несут практической ценности для построения моделей машинного обучения.

## 2.3 Приведение числовых типов

```python
numeric_cols_carbon = carbon_df.columns.drop(['SAE Grade', 'Conditions'])
numeric_cols_stainless = stainless_df.columns.drop(['SAE Grade', 'Conditions'])

carbon_df[numeric_cols_carbon] = carbon_df[numeric_cols_carbon].apply(
    pd.to_numeric, errors='coerce'
)

stainless_df[numeric_cols_stainless] = stainless_df[numeric_cols_stainless].apply(
    pd.to_numeric, errors='coerce'
)
```

В рамках этого шага выполняется приведение всех количественных признаков к числовому формату с использованием pd.to_numeric. Некорректные значения автоматически заменяются на NaN, что упрощает последующую работу с пропущенными данными.

### Преобразование Min / Max -> среднее значение

Для упрощения модели и снижения размерности данных каждый диапазон заменяется на среднее значение концентрации элемента.

```python
elements = ['C', 'Mn', 'Si', 'Ni', 'Cr', 'Mo', 'Ti']

for el in elements:
    carbon_df[el] = carbon_df[[f'{el} (Min)', f'{el} (Max)']].mean(axis=1)

carbon_df[
    ['SAE Grade', 'Conditions'] +
    [f'{el} (Min)' for el in elements] +
    [f'{el} (Max)' for el in elements] +
    elements
].head()
```

```python
elements = ['C', 'Mn', 'Si', 'Ni', 'Cr', 'Mo', 'Ti']

for el in elements:
    stainless_df[el] = stainless_df[[f'{el} (Min)', f'{el} (Max)']].mean(axis=1)

stainless_df['N'] = pd.to_numeric(stainless_df['N'], errors='coerce')

stainless_df[
    ['SAE Grade', 'Conditions'] +
    [f'{el} (Min)' for el in elements] +
    [f'{el} (Max)' for el in elements] +
    elements +
    ['N']
].head()
```

После преобразования для каждого элемента добавляется новый столбец со средним значением, при этом исходные столбцы Min и Max временно сохраняются для визуальной проверки корректности вычислений.  
На данном этапе можно убедиться, что усреднение выполнено корректно и не приводит к искажениям данных.

# 4 Анализ корреляций

## 4.1 Корреляции для углеродистых сталей

```python
carbon_features = [
    'C', 'Mn',
    'UTS (MPa)', 'YS (MPa)',
    'Elongation (%)', 'Hardness (HB)'
]
```

### Корреляционная матрица

```python
carbon_corr = carbon_df[carbon_features].corr()
carbon_corr
```

Содержание углерода (C) имеет умеренно положительную корреляцию с:
- пределом прочности (UTS, r ≈ 0.51),
- пределом текучести (YS, r ≈ 0.35),
- твердостью (Hardness, r ≈ 0.55).
  
Это соответствует физической природе сталей: увеличение содержания углерода приводит к росту прочности и твёрдости за счёт упрочнения структуры.  
Марганец (Mn) демонстрирует слабую положительную корреляцию с прочностными характеристиками, что указывает на его вспомогательную роль по сравнению с углеродом.

Углерод имеет выраженную отрицательную корреляцию с удлинением (r ≈ -0.58), что отражает классический компромисс между прочностью и пластичностью.  
Аналогичная отрицательная связь наблюдается между:
- UTS и Elongation,
- YS и Elongation,
- Hardness и Elongation.
  
Это подтверждает, что рост прочностных характеристик сопровождается снижением пластичности материала.

Между UTS и YS наблюдается очень высокая корреляция (r ≈ 0.83).  
Между UTS и твердостью - почти линейная зависимость (r ≈ 0.93).  
Между YS и твердостью - также высокая корреляция (r ≈ 0.87).  
Это указывает на сильную взаимосвязанность целевых переменных, что важно учитывать при построении многозадачных моделей.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure()
plt.imshow(carbon_corr)
plt.colorbar()
plt.xticks(range(len(carbon_features)), carbon_features, rotation=45)
plt.yticks(range(len(carbon_features)), carbon_features)
plt.title('Корреляционная матрица для углеродистых сталей')
plt.tight_layout()
plt.show()
```
