## 2. Предварительная обработка данных

### 2.1 Исследование пропущенных значений

```python
carbon_df.isna().sum().sort_values(ascending=False)
```

```python
stainless_df.isna().sum().sort_values(ascending=False)
```

На данном этапе определяется количество пропусков по каждому признаку. Результаты используются для выбора стратегии очистки данных и подготовки признакового пространства к дальнейшему анализу.

### 2.2 Удаление наблюдений без целевой переменной

```python
stainless_df = stainless_df[
    stainless_df['UTS (MPa)'].notna()
]

carbon_df = carbon_df[
    carbon_df['UTS (MPa)'].notna()
]
```

Для последующего моделирования необходимы корректные значения целевых характеристик. Записи без значения предела прочности при растяжении исключаются из анализа.

### 2.3 Преобразование типов данных

```python
numeric_cols_carbon = carbon_df.columns.drop(['SAE Grade', 'Conditions'])
numeric_cols_stainless = stainless_df.columns.drop(['SAE Grade', 'Conditions'])

carbon_df[numeric_cols_carbon] = carbon_df[numeric_cols_carbon].apply(
    pd.to_numeric, errors='coerce'
)

stainless_df[numeric_cols_stainless] = stainless_df[numeric_cols_stainless].apply(
    pd.to_numeric, errors='coerce'
)
```

Все количественные признаки приводятся к числовому формату. Это обеспечивает корректную работу статистических методов, визуализаций и алгоритмов машинного обучения.

#### Формирование усреднённых значений химического состава

Для элементов, представленных диапазонами минимальных и максимальных значений, рассчитывается среднее значение концентрации. Такой подход позволяет уменьшить количество признаков и упростить дальнейший анализ.

```python
elements = ['C', 'Mn', 'Si', 'Ni', 'Cr', 'Mo', 'Ti']

for el in elements:
    carbon_df[el] = carbon_df[[f'{el} (Min)', f'{el} (Max)']].mean(axis=1)

carbon_df[
    ['SAE Grade', 'Conditions'] +
    [f'{el} (Min)' for el in elements] +
    [f'{el} (Max)' for el in elements] +
    elements
].head()
```

```python
elements = ['C', 'Mn', 'Si', 'Ni', 'Cr', 'Mo', 'Ti']

for el in elements:
    stainless_df[el] = stainless_df[[f'{el} (Min)', f'{el} (Max)']].mean(axis=1)

stainless_df['N'] = pd.to_numeric(stainless_df['N'], errors='coerce')

stainless_df[
    ['SAE Grade', 'Conditions'] +
    [f'{el} (Min)' for el in elements] +
    [f'{el} (Max)' for el in elements] +
    elements +
    ['N']
].head()
```

Промежуточная проверка позволяет убедиться в корректности расчёта новых признаков и согласованности полученных значений с исходными данными.

## 5. Формирование наборов данных для моделирования

```python
# ДОБАВЛЕНИЕ КАТЕГОРИАЛЬНЫХ ПРИЗНАКОВ
import pandas as pd

# 1. Кодирование Conditions (термическая обработка) для углеродистых сталей
carbon_df['Conditions_encoded'] = carbon_df['Conditions'].astype('category').cat.codes

# 2. Кодирование SAE Grade (группа марок) - берём первые 2 цифры как класс стали
carbon_df['SAE_Class'] = carbon_df['SAE Grade'].astype(str).str[:2].astype('category').cat.codes

# 3. Для нержавеющих сталей - аналогично
stainless_df['Conditions_encoded'] = stainless_df['Conditions'].astype('category').cat.codes
stainless_df['SAE_Class'] = stainless_df['SAE Grade'].astype(str).str.extract(r'(\d+)')[0].fillna('0').astype(int)
stainless_df['SAE_Class'] = stainless_df['SAE_Class'].astype('category').cat.codes

# 4. Добавляем в списки признаков
carbon_features_extended = ['C', 'Mn', 'Conditions_encoded', 'SAE_Class']
stainless_features_extended = ['C', 'Mn', 'Si', 'Ni', 'Cr', 'Conditions_encoded', 'SAE_Class']

# 5. Обновляем финальные датасеты
carbon_df_prepared = carbon_df[carbon_features_extended + target_cols].dropna()
stainless_df_prepared = stainless_df[stainless_features_extended + target_cols].dropna()
```

```python
carbon_df_prepared.to_csv(
    "data/carbon_steel_prepared.csv",
    index=False
)

stainless_df_prepared.to_csv(
    "data/stainless_steel_prepared.csv",
    index=False
)
```

На основании выполненного анализа формируются итоговые наборы признаков, которые будут использоваться при построении и сравнении моделей машинного обучения.
